```python
import pandas as pd
import plotly.express as px
from pathlib import Path

from UNRegion_Choropleth_Visualiser import create_un_region_choropleth

cwd = Path.cwd()

# Input paths
results_folder =  cwd.parent / "results"
file_path_EssentialWorkersByCountry = results_folder / "EssentialWorkersByCountry.csv"
file_path_EssentialWorkersByRegion = results_folder / "EssentialWorkersByRegion.csv"

# Output paths
IndoorVitalWorkersByRegion_PATH = results_folder / "IndoorVitalWorkersByRegion.html"
IndoorEssentialWorkersByRegion_PATH = results_folder / "IndoorEssentialWorkersByRegion.html"
IndoorVitalWorkersByCountry_PATH = results_folder / "IndoorVitalWorkersByCountry.html"
IndoorEssentialWorkersByCountry_PATH = results_folder / "IndoorEssentialWorkersByCountry.html"
```

### Create Region plots

```python
df_region = pd.read_csv(file_path_EssentialWorkersByRegion)
IndoorVitalWorkersByRegion = create_un_region_choropleth(df_region, region_col="Region", value_col="%Indoor Vital Workers", title="%Indoor Vital Workers by Region", color_scale="Greens")
IndoorEssentialWorkersByRegion = create_un_region_choropleth(df_region, region_col="Region", value_col="%Indoor Essential Workers", title="%Indoor Essential Workers by Region", color_scale="Greens")
```

### Create Country plots

```python
# Load your data
df_country = pd.read_csv(file_path_EssentialWorkersByCountry)

# First map: Indoor Essential Workers
IndoorEssentialWorkersByCountry = px.choropleth(
    df_country,
    locations="Country Code",
    locationmode="ISO-3",
    color="%Indoor Essential Workers",
    hover_name="Country Code",
    color_continuous_scale='spectral',
    title="Indoor Essential Workers by Country"
)

# Second map: Indoor Vital Workers
IndoorVitalWorkersByCountry = px.choropleth(
    df_country,
    locations="Country Code",
    locationmode="ISO-3",
    color="%Indoor Vital Workers",
    hover_name="Country Code",
    color_continuous_scale='spectral',
    title="Indoor Vital Workers by Country"
)
```

### Save plots

```python
IndoorVitalWorkersByRegion_PATH = results_folder / "IndoorVitalWorkersByRegion.html"
IndoorEssentialWorkersByRegion_PATH = results_folder / "IndoorEssentialWorkersByRegion.html"
IndoorVitalWorkersByCountry_PATH = results_folder / "IndoorVitalWorkersByCountry.html"
IndoorEssentialWorkersByCountry_PATH = results_folder / "IndoorEssentialWorkersByCountry.html"


IndoorVitalWorkersByRegion.write_html(IndoorVitalWorkersByRegion_PATH)
IndoorEssentialWorkersByRegion.write_html(IndoorEssentialWorkersByRegion_PATH)
IndoorVitalWorkersByCountry.write_html(IndoorVitalWorkersByCountry_PATH)
IndoorEssentialWorkersByCountry.write_html(IndoorEssentialWorkersByCountry_PATH)
```
